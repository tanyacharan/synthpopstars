CTGANSynthesizer for Finance Data

```python
!pip install -q sdv pandas scikit-learn matplotlib seaborn

import pandas as pd
from google.colab import files
from sdv.single_table import CTGANSynthesizer
from sdv.metadata import SingleTableMetadata
import warnings
warnings.filterwarnings('ignore')

# Upload file
print("1. Upload your german.data file")
uploaded = files.upload()

# Load dataset (space-separated, no headers)
print("2. Reading german.data...")
column_names = [
    "checking_status", "duration", "credit_history", "purpose",
    "credit_amount", "savings_status", "employment",
    "installment_commitment", "personal_status", "other_parties",
    "residence_since", "property_magnitude", "age",
    "other_payment_plans", "housing", "existing_credits",
    "job", "num_dependents", "own_telephone", "foreign_worker",
    "class"
]

df_real = pd.read_csv("german.data", sep=' ', header=None, names=column_names)

print("Data shape:", df_real.shape)

# Convert categorical columns to object
for col in df_real.columns:
    if df_real[col].dtype == 'object':
        df_real[col] = df_real[col].astype('object')

# Detect metadata
print("3. Detecting metadata...")
metadata = SingleTableMetadata()
metadata.detect_from_dataframe(df_real)

# Train CTGAN (1 generation only)
print("4. Training CTGAN...")
synthesizer = CTGANSynthesizer(metadata, epochs=10, verbose=False)
synthesizer.fit(df_real)

# Generate synthetic dataset
print("5. Generating synthetic data...")
df_synth = synthesizer.sample(num_rows=len(df_real))

# Save to CSV
output_filename = "german_credit_synthetic_gen1.csv"
df_synth.to_csv(output_filename, index=False)

print(f"6. Saved synthetic data to {output_filename}")

# Download
files.download(output_filename)

print("Done.")
```

GaussianCopulaSynthesizer for Healthcare Data

```python
!pip install -q sdv pandas scikit-learn matplotlib seaborn

import pandas as pd
from google.colab import files
from sdv.single_table import GaussianCopulaSynthesizer
from sdv.metadata import SingleTableMetadata
import warnings
warnings.filterwarnings('ignore')

print("1. Upload your diabetic_data.csv file")
uploaded = files.upload()

print("2. Reading diabetic_data.csv...")
df_real = pd.read_csv("diabetic_data.csv")

# Replace ? with missing values
df_real = df_real.replace("?", pd.NA)

print("Data shape:", df_real.shape)

# Keep object columns as categorical-like strings
for col in df_real.columns:
    if df_real[col].dtype == 'object':
        df_real[col] = df_real[col].astype('object')

print("3. Detecting metadata...")
metadata = SingleTableMetadata()
metadata.detect_from_dataframe(df_real)

print("4. Training Gaussian Copula synthesizer...")
synthesizer = GaussianCopulaSynthesizer(metadata)
synthesizer.fit(df_real)

print("5. Generating one synthetic dataset...")
df_synth = synthesizer.sample(num_rows=len(df_real))

output_filename = "diabetic_data_synthetic_gen1.csv"
df_synth.to_csv(output_filename, index=False)

print(f"6. Saved synthetic data to {output_filename}")
files.download(output_filename)

print("Done.")
```
